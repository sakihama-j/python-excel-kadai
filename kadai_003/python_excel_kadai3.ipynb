```python
import pandas as pd

# ファイルの読み込み
df_2022 = pd.read_excel('2022_年間売上表.xlsx')
df_2023 = pd.read_excel('2023_年間売上表.xlsx')

# データの連結
df_concat = pd.concat([df_2022, df_2023], ignore_index=True)

# グループ化して合計を算出
# 「金額（千円）」列だけを集計対象にする
grouped = df_concat.groupby(['商品', '売上年'])['金額（千円）'].sum()

# Excelファイルに書き出す
grouped.to_excel('売上集計表.xlsx')

import openpyxl
from openpyxl.styles import PatternFill
# ワークブックの読み込み
wb = openpyxl.load_workbook('売上集計表.xlsx')
ws = wb.active

# 背景色の設定を作成
fill_settings = PatternFill(patternType='solid', fgColor='F2F2F2')

# 1行目のすべてのセルに対して処理を行う
for cell in ws[7]:
    cell.fill = fill_settings

# ファイルを保存
wb.save('売上集計表.xlsx')

```
